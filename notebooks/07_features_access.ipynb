# Part 7 — Market-access feature

**Source:** `Oxford/MAP/accessibility_to_cities_2015_v1_0`

**Output:** `feat_access` (250 m). **DoD:** low near cities, high in remote NE Goiás.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import ee, geemap
import utils, features
project = utils.init()
print('EE initialized; project =', project)
```

```python
aoi = utils.load_aoi(project)
print('AOI area (km^2):', round(aoi.area(1000).divide(1e6).getInfo(), 1))
```

### Build the feature image

```python
img = features.access_features(aoi)
print('bands:', img.bandNames().getInfo())
```

### DoD sanity — per-band min/mean/max

```python
utils.range_report(img, aoi)
```

### Quick look

```python
Map = geemap.Map()
Map.centerObject(aoi, 7)
Map.addLayer(img.select('access_logtt'), {'min': 0, 'max': 7, 'palette': ['green','yellow','red']}, 'access_logtt')
Map.addLayer(aoi, {}, 'AOI', False)
Map
```

### Export to EE asset

```python
utils.ensure_folder(project)
task = utils.export_image(img, project, 'feat_access', aoi)
print('export', task.status()['description'], '->', task.status()['state'])
```
