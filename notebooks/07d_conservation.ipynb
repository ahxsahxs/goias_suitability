# Part 7d — Conservation-value features (NOT stacked)

**Source:** `WCMC/WDPA` reserve distance + SRTM ruggedness + `NASA/ORNL/biomass_carbon_density`

**Output:** `feat_conservation` (250 m). **DoD:** cv_pa_dist 0 inside reserves; cv_ruggedness ~m-scale; cv_carbon Mg C/ha. Consumed by conservation membership via `cv_` routing; deliberately NOT in the stack.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import ee, geemap
import utils, features
project = utils.init()
print('EE initialized; project =', project)
```

```python
aoi = utils.load_aoi(project)
print('AOI area (km^2):', round(aoi.area(1000).divide(1e6).getInfo(), 1))
```

### Build the feature image

```python
img = features.conservation_features(aoi)
print('bands:', img.bandNames().getInfo())
```

### DoD sanity — per-band min/mean/max

```python
utils.range_report(img, aoi)
```

### Quick look

```python
Map = geemap.Map()
Map.centerObject(aoi, 7)
Map.addLayer(img.select('cv_carbon'), {'min': 0, 'max': 60, 'palette': ['white','darkgreen']}, 'cv_carbon')
Map.addLayer(aoi, {}, 'AOI', False)
Map
```

### Export to EE asset

```python
utils.ensure_folder(project)
task = utils.export_image(img, project, 'feat_conservation', aoi)
print('export', task.status()['description'], '->', task.status()['state'])
```
