# Part 2 — Climate & water-balance features

**Source:** `IDAHO_EPSCOR/TERRACLIMATE` monthly climatology, 1991–2020

**Output:** `feat_climate` (250 m). **DoD:** band ranges sane (e.g. annual precip ~800–2000 mm; aridity 0–2).

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import ee, geemap
import utils, features
project = utils.init()
print('EE initialized; project =', project)
```

```python
aoi = utils.load_aoi(project)
print('AOI area (km^2):', round(aoi.area(1000).divide(1e6).getInfo(), 1))
```

### Build the feature image

```python
img = features.climate_features(aoi)
print('bands:', img.bandNames().getInfo())
```

### DoD sanity — per-band min/mean/max

```python
utils.range_report(img, aoi)
```

### Quick look

```python
Map = geemap.Map()
Map.centerObject(aoi, 7)
Map.addLayer(img.select('clim_pr_annual'), {'min': 800, 'max': 2000, 'palette': ['red','yellow','blue']}, 'clim_pr_annual')
Map.addLayer(aoi, {}, 'AOI', False)
Map
```

### Export to EE asset

```python
utils.ensure_folder(project)
task = utils.export_image(img, project, 'feat_climate', aoi)
print('export', task.status()['description'], '->', task.status()['state'])
```
