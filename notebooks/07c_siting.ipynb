# Part 7c — Siting features (solar/pisci; NOT stacked)

**Source:** TerraClimate srad / FAO-56 extraterrestrial Ra (clearness) + `JRC/GSW1_4` seasonal water

**Output:** `feat_siting` (250 m). **DoD:** clearness ~0.45-0.55; seasonal-water distance far below permanent-water distance. Consumed by solar/pisci membership via `sit_` routing; deliberately NOT in the stack.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import ee, geemap
import utils, features
project = utils.init()
print('EE initialized; project =', project)
```

```python
aoi = utils.load_aoi(project)
print('AOI area (km^2):', round(aoi.area(1000).divide(1e6).getInfo(), 1))
```

### Build the feature image

```python
img = features.siting_features(aoi)
print('bands:', img.bandNames().getInfo())
```

### DoD sanity — per-band min/mean/max

```python
utils.range_report(img, aoi)
```

### Quick look

```python
Map = geemap.Map()
Map.centerObject(aoi, 7)
Map.addLayer(img.select('sit_clearness'), {'min': 0.45, 'max': 0.52, 'palette': ['white','orange']}, 'sit_clearness')
Map.addLayer(aoi, {}, 'AOI', False)
Map
```

### Export to EE asset

```python
utils.ensure_folder(project)
task = utils.export_image(img, project, 'feat_siting', aoi)
print('export', task.status()['description'], '->', task.status()['state'])
```
