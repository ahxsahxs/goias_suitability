# Part 7b — Generic land-cover mask (ESA WorldCover)

**Source:** `ESA/WorldCover/v200` (10 m, 2021), aggregated to 250 m

**Output:** `feat_landcover` (250 m). **DoD:** built-up around Goiânia/Brasília and major reservoirs excluded.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import ee, geemap
import utils, features
project = utils.init()
print('EE initialized; project =', project)
```

```python
aoi = utils.load_aoi(project)
print('AOI area (km^2):', round(aoi.area(1000).divide(1e6).getInfo(), 1))
```

### Build the feature image

```python
img = features.landcover_features(aoi)
print('bands:', img.bandNames().getInfo())
```

### DoD sanity — per-band min/mean/max

```python
utils.range_report(img, aoi)
```

### Quick look

```python
Map = geemap.Map()
Map.centerObject(aoi, 7)
Map.addLayer(img.select('mask_excluded'), {'min': 0, 'max': 1, 'palette': ['white','black']}, 'mask_excluded')
Map.addLayer(aoi, {}, 'AOI', False)
Map
```

### Export to EE asset

```python
utils.ensure_folder(project)
task = utils.export_image(img, project, 'feat_landcover', aoi)
print('export', task.status()['description'], '->', task.status()['state'])
```
