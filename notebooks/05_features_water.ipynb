# Part 5 — Hydrology & water-access features

**Source:** `JRC/GSW1_4/GlobalSurfaceWater` + `WWF/HydroSHEDS/15ACC`

**Output:** `feat_water` (250 m). **DoD:** distance raster continuous; reservoirs present.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import ee, geemap
import utils, features
project = utils.init()
print('EE initialized; project =', project)
```

```python
aoi = utils.load_aoi(project)
print('AOI area (km^2):', round(aoi.area(1000).divide(1e6).getInfo(), 1))
```

### Build the feature image

```python
img = features.water_features(aoi)
print('bands:', img.bandNames().getInfo())
```

### DoD sanity — per-band min/mean/max

```python
utils.range_report(img, aoi)
```

### Quick look

```python
Map = geemap.Map()
Map.centerObject(aoi, 7)
Map.addLayer(img.select('water_dist'), {'min': 0, 'max': 20000, 'palette': ['blue','white','brown']}, 'water_dist')
Map.addLayer(aoi, {}, 'AOI', False)
Map
```

### Export to EE asset

```python
utils.ensure_folder(project)
task = utils.export_image(img, project, 'feat_water', aoi)
print('export', task.status()['description'], '->', task.status()['state'])
```
