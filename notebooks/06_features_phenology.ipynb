# Part 6 — Vegetation phenology features

**Source:** `MODIS/061/MOD13Q1` NDVI, harmonic fit 2001–2020

**Output:** `feat_phenology` (250 m). **DoD:** amplitude high in croplands, low in dense native veg.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
import ee, geemap
import utils, features
project = utils.init()
print('EE initialized; project =', project)
```

```python
aoi = utils.load_aoi(project)
print('AOI area (km^2):', round(aoi.area(1000).divide(1e6).getInfo(), 1))
```

### Build the feature image

```python
img = features.phenology_features(aoi)
print('bands:', img.bandNames().getInfo())
```

### DoD sanity — per-band min/mean/max

```python
utils.range_report(img, aoi)
```

### Quick look

```python
Map = geemap.Map()
Map.centerObject(aoi, 7)
Map.addLayer(img.select('phen_amplitude'), {'min': 0, 'max': 0.4, 'palette': ['white','green']}, 'phen_amplitude')
Map.addLayer(aoi, {}, 'AOI', False)
Map
```

### Export to EE asset

```python
utils.ensure_folder(project)
task = utils.export_image(img, project, 'feat_phenology', aoi)
print('export', task.status()['description'], '->', task.status()['state'])
```
